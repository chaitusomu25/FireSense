```python
#operations in tuple
t1=(10,20,30,40)
t2=(50,60,70,80)
print(t1+t2)   #concatination
for i in t1:
  print(i)     #iteration
print(t2*4)    #repetition
print(t1[0:2]) #slicing
print(t2[::-1])  #reversing
```

```python
#functions in tuple
p=(3,11,16,65,10,15,20,25,30)
print(sum(p))
print(max(p))
print(min(p))
print(sorted(p))
```

```python
#methods in tuple
m=(25,12,2,7,22,2,30,6,24,5,2)
print(m.count(2))
print(m.index(7))
```

```python
#zipping
a1=(1,3,5,7,9)
a2=(2,4,6,8,10)
print(tuple(zip(a1,a2)))
```

```python
#creating tuple
b=tuple(map(int,input("enter the elements").split()))
print(b)
```

```python
#tuple comprehension
c=tuple(int(i) for i in input().split())
print(c)
```

write a program to count the number of vowels in a string(no control flow allowed)
